```python
# Split Dataset
```

```python
x = df2.drop("student_marks",axis = "columns")
y = df2.drop("study_hours",axis = "columns")
print("Shape of X = " , x.shape)
print("Shape of Y = " , y.shape)
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state=0)
print("Shape 0f X_Train = ",x_train.shape)
print("Shape 0f X_Test  = ",x_test.shape)
print("Shape 0f Y_Train = ",y_train.shape)
print("Shape 0f Y_Train = ",y_test.shape)
```

# Select a Model and Train it

```python
# Y = M*X + C
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
```

```python
lr.fit(x_train, y_train)
```

```python
lr.get_params()
```

```python
m = lr.coef_  # M
print(m)
```

```python
c = lr.intercept_  # C
print(c)
```

```python
y = m * 10 + c
print(y)
```

```python
y = m * 11 + c
print(y)
```

```python
lr.predict([[11]]).round(2)
```

```python
lr
```

```python
y_pred = lr.predict(x_test)
y_pred
```

```python
# np.c_  means combine things column wise
pd.DataFrame(np.c_[x_test, y_test, y_pred],columns = ["Study_Hours","Student_Marks_Orginal",
                                                      "Student_Marks_Predicted"])
```

## Fine_tune Model

```python
lr.score(x_test,y_test) # Variance Score
```

```python
lr.score(x_train,y_train) # Bias Score
```
